# OOP Concepts

-**Encapsulation**: Bundling data and methods within a class, hiding internal implementation.

-**Inheritance**: Defining a new class (child) based on an existing class (parent) to reuse code and form hierarchies.

-**Polymorphism**: Allowing objects of different classes to be treated through a common interface (often via overridden methods or duck typing).

-**Abstraction**: Using abstract base classes or interfaces to separate the conceptual model from concrete implementation.

# use Of One UnderScore (_variable) meaning its protected by naming convention only

```python
class BankAccount:
  def __init__(self,owner,balance):
    self.owner=owner
    self._balance=balance#Private variable by name convection only
  def show_balance(self):
    return f'{self.owner} balance is {self._balance}'
  def deposit(self,amount):
    self._balance+=amount
```

```python
account=BankAccount('Hamza',100000)

print(account.show_balance())
account.deposit(50000)
print(account.show_balance())
#This means its protected my naming convetion only and can be accessed
```

```python
#_balance wasn’t protected from outside misuse.
#You could easily change it, even to an unrealistic value.
account._balance=-25000000
print(account.show_balance())
```

# 2.Inheritance

Inheritance lets you create a new class (called child or subclass) based on an existing class (called parent or superclass).

# single inheritance

```python
class Animal:
  def speak(self):
    print('Animal sound Initialized')
class Dog(Animal):#Dog inherits from Animal
  def speak(self):
    print('WOOF!')
dog=Dog()
dog.speak()
```

# Method Overriding and super()

**Method Overriding** is when a child class provides its own version of a method that is already defined in the parent class.
overriding=Behaviour Upgrade


```python
class Animal:
  def sound(self):
    print('Animal sound Initialized')
class Cat(Animal):#This is method overriding which is a child class own version of an already defined parent method
  def sound(self):
    print('MEOW!')
cat=Cat()
cat.sound()
```

```python
class Company:
  def ceo(self):
    print('lets find out the ceos of companies')
class Microsoft(Company):
  def ceo(self):
    print('Bill Gates')
microsoft=Microsoft()
microsoft.ceo()
```

# Super() Function

The super() function is used to call a method from the parent class.

```python
#Using super() function in constructors
class Animal:
  def __init__(self,name):
    self.name=name

class Dog(Animal):
  def __init__(self, name,breed):
    super().__init__(name)
    self.breed=breed
dog=Dog('Zahra','Rottweiler')
print(dog.name)
print(dog.breed)
```

```python
#Using Super() function in Overriding Method
class Animal:
  def sound(self):
    print('Animal sound Intialized')
class Cat(Animal):#This is method overriding which is a child class own version of an already defined parent method
  def sound(self):
    super().sound()
    print('MEOW!')

cat=Cat()
cat.sound()
```

# Polymorphism And Duck Typing

**Polymorphism** means "many forms" — different classes can have methods with the same name, but their behavior depends on the object calling them.

**Real-Life Analogy:**

A drive() method means something different for:

A Car — rolls wheels

A Boat — moves on water

A Plane — takes off

They all respond to drive() — but in their own way.

```python
class Car:
  def drive(self):
    print('A Car rolls wheels ')
class Boat:
  def drive(self):
    print('A Boat moves on water')
class Plane:
  def drive(self):
    print('A Plane takes Off')

def drive_method(drive_object):
  drive_object.drive()

car=Car()
boat=Boat()
plane=Plane()

drive_method(car)
drive_method(boat)
drive_method(plane)
```

```python
class Dog:
  def sound(self):
    return 'Woof!'
class Cat:
  def sound(self):
    return 'Meow!'

def animal_sound(animal):
  print(animal.sound())

dog=Dog()
cat=Cat()

animal_sound(dog)
animal_sound(cat)
```

# Duck Typing

Duck Typing is Python's style of polymorphism.
You don’t care what the object’s actual type is, as long as it behaves correctly.

its from the analogy “If it walks like a duck and quacks like a duck, it’s a duck.”



```python
class Duck:
  def quack(self):
    print('QuackQuack')
class Human:
  def quack(self):
    print('Am a Human pretending to be a Duck but giving out the sound')

def anything_that_quacks(quacking_object):
  quacking_object.quack()

duck=Duck()
human=Human()

anything_that_quacks(duck)
anything_that_quacks(human)
#	"Type" is based on behavior, not class name	and its built in
```

# Abstraction and Interfaces
Abstraction means hiding the complex details and showing only the essential features of an object.

Python uses the **abc** module to define abstract classes and abstract methods.

-**Abstract Class:** A class that can’t be instantiated and is meant to be a blueprint.

-**Abstract Method:** A method declared, but must be implemented by any subclass.



```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass  # Abstract method — no implementation here

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

# animal = Animal()  # ERROR: Can't instantiate abstract class

dog = Dog()
cat = Cat()

print(dog.speak())  # Woof!
print(cat.speak())  # Meow!
```

# Interfaces
Interfaces are abstract classes with only abstract methods.

Any class inheriting from the interface must implement all those methods.



```python
from abc import ABC, abstractmethod

class Flyable(ABC):
    @abstractmethod
    def fly(self):
        pass

class Bird(Flyable):
    def fly(self):
        print("Bird is flying")

class Airplane(Flyable):
    def fly(self):
        print("Airplane is flying")

# f = Flyable()  # ERROR: Can't instantiate interface

b = Bird()
a = Airplane()

b.fly()  # Bird is flying
a.fly()  # Airplane is flying
```

```python
from abc import ABC,abstractmethod

class Rideable(ABC):
  def ride_object(self):
    pass
class Bike(Rideable):
  def ride_object(self):
    return 'A Bike is Rideable'
class Horse(Rideable):
  def ride_object(self):
    return 'A Horse is Rideable'

bike=Bike()
horse=Horse()

print(bike.ride_object())
print(horse.ride_object())
```

# Special Dunder Method

eg __ init __ -used as a constructor

```python
#__str__ -Human Readable String
class Person:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Person: {self.name}"

p = Person("Hamza")
print(p)
```

```python
#__repr__ -Official string used in debugging
class Person:
    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return f"Person('{self.name}')"

p = Person("Hamza")
print(repr(p))
```

```python
#__len__ – Called by len()
class Basket:
    def __init__(self, items):
        self.items = items

    def __len__(self):
        return len(self.items)

b = Basket(['apple', 'banana'])
print(len(b))
```

```python
#__add__-Arithmetic Operator
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):
        return Money(self.amount + other.amount)

    def __str__(self):
        return f"${self.amount}"

m1 = Money(100)
m2 = Money(50)
print(m1 + m2)  # $150
```

```python
#rock paper scissors game
import random

print("Welcome to the Rock Paper Scissors Game")
possiblities = ["rock", "paper", "scissors"]

pc_score = 0
user_score = 0

for i in range(3):
  pc_choice = random.choice(possiblities)

  user_choice = input("Enter your choice (Rock, Paper or Scissors) : ").lower()

  if user_choice == pc_choice:
    print(f"It's a tie. You both chose {user_choice}")
  elif user_choice == "rock" and pc_choice == "scissors" or user_choice == "paper" and pc_choice == "rock" or user_choice == "scissors" and pc_choice == "paper":
    print(f"You win : PC chose {pc_choice} and you chose {user_choice}")
    user_score += 1
  else:
    print(f"You lose : PC chose {pc_choice} and you chose {user_choice}")
    pc_score += 1

if user_score == pc_score:
  print(f"It's a tie. You both had {pc_score} points")
elif user_score > pc_score:
  print(f"You win. You had {user_score} while PC had {pc_score}")
else:
  print(f"You lose. You had {user_score} while PC had {pc_score}")
```
